## Exercise 2 - Filtering

Return the following subsets of the dataframe.

1. How many listings are there with a price less than 100? 


2. Find how many listings there are in just Brooklyn.


3. Find how many listings there are in Brooklyn with a price less than 100.


4. Using `.isin()` select anyone that has the host name of Michael, David, John, and Daniel.


5. Create a new column called `adjusted_price` that has $100 added to every listing in Williamsburg.  The prices for all other listings should be the same as the were before. 


6. What % of the rooms are private, and what % of the rooms are shared.  
    * Hint, use `.value_counts()`


```python
# 1. How many listings are there with a price less than 100? 
under100 = df['price'] < 100
under100_value = df[under100].copy()
under100_value.head()
```

```python
# 2. Make a new DataFrame of listings in Brooklyn named `df_bk` 
# and find how many listings in just Brooklyn.
brooklyn = df['neighbourhood_group'] == 'Brooklyn'
df_bk = df[brooklyn].copy()
df_bk.head()
```

```python
# 3. Find how many listings there are in Brooklyn with a price less than 100.
brooklynLess100 = df_bk['price'] < 100
df_bk_100 = df_bk[brooklynLess100].copy()
df_bk_100.neighbourhood_group.value_counts()


```

```python
# 4. Using `.isin()` select anyone that has the host name of Michael, David, John, and Daniel.
# How many total are there that have those names
name = ['Michael' , 'David' , 'John' , 'Daniel']
condition = df.host_name.isin(name)

df[condition]['host_name'].value_counts()
```

```python
# 5. Create a new column called `adjusted_price` that has $100 added to every listing in Williamsburg.  
# The prices for all other listings should be the same as the were before. 
c1 = df['neighbourhood'] == 'Williamsburg'
df['adjusted_price'] = np.where(c1, df['price'] + 100, df['price'])
df.head()
```

```python
# 6. What % of the rooms are private, and what % of the rooms are shared.  
room_types = df['room_type'].value_counts()
room_types_percent = df['room_type'].value_counts(normalize=True) * 100
room_types_percent
```

# Exercise 3 - Grouping

1. Using `groupby`, count how many listings are in each neighbourhood_group.


2. Using `groupby`, find the mean price for each of the neighbourhood_groups. 


3. Using `groupby` and `.agg()`, find the min and max price for each of the neighbourhood_groups. 


4. Using `groupby`, find the median price for each room type in each neighbourhood_group.


5. Using `groupby` and `.agg()`, find the count, min, max, mean, median, and std of the prices for each room type in each neighbourhood_group.

```python
# 1. Using `groupby`, count how many listings are in each neighbourhood_group.
nb_groups = df.groupby('neighbourhood_group')
nb_counts = nb_groups.size()
nb_counts
```

```python
# 2. Using `groupby`, find the mean price for each of the neighbourhood_groups. 
nb_price = df.groupby('neighbourhood_group')
nb_price['price'].mean()

```

```python
# 2.5. Using `groupby`, find the mean price for each room_type. 
rm_type = df.groupby('room_type')
rm_type['price'].mean()
```

```python
# 3. Using `groupby` and `.agg()`, find the min and max price for each of the neighbourhood_groups. 
nb_p = df.groupby('neighbourhood_group')
nb_p['price'].agg( ['min', 'max'] )

```

```python
# 4. Using `groupby`, find the mean price for each room_type in each neighbourhood_group.

rm_nb = df.groupby(['room_type', 'neighbourhood_group'])
rm_nb['price'].mean()
```

```python
# 5. Using `groupby` and `.agg()`, find the count, min, max, mean, median, and std of the prices 
# for each room type in each neighbourhood_group.

rm_nb_agg = df.groupby(['room_type', 'neighbourhood_group'])
rm_nb_agg['price'].agg(['count', 'min', 'max', 'mean', 'median', 'std'])

```

```python
# 2. Do join that keeps all the records for each table.
data = pd.merge(
    df_prices,
    df_n_listings,
    on='neighbourhood_group', 
    how='outer'
)
data.head()
```

```python
import pandas as pd

df = pd.read_csv('../data/grammys.csv')

# 1. Album of the Year in 2016
result1 = df[
    (df['year'] == 2016) &
    (df['category'] == 'Album of the Year') &
    (df['winner'] == True)
]

print(result1[['year', 'category', 'nominee', 'workers']])

# 2. Best Rap Album in 2009
result2 = df[
    (df['year'] == 2009) &
    (df['category'] == 'Best Rap Album') &
    (df['winner'] == True)
]
print(result2[['year', 'category', 'nominee', 'workers']])

# 3. Kendrick Lamar nominations and wins
k = df[df['workers'].str.contains('Kendrick Lamar', na=False, case=False)]

print('Total nominations:', len(k))
print('Total wins:', k['winner'].sum())
print(k[['year', 'category', 'nominee', 'winner']].head(20))
```

### (Extra Hard) Download the files for the past 5 years into a new folder and add them all into one data frame using pandas, then save that new file.
* Try to not hard code in the file names. We want to do this programmatically because what if we want to add new/more file names in the future.

```python
import pandas as pd
from pathlib import Path

folder = Path('../data')   # or wherever your game log files live
all_files = sorted(folder.glob('*.csv'))

df_list = []
for f in all_files:
    temp = pd.read_csv(f, header=None)
    # rename columns here if needed
    df_list.append(temp)

combined = pd.concat(df_list, ignore_index=True)
combined.to_csv('../data/all_game_logs_2020_2024.csv', index=False)
```
